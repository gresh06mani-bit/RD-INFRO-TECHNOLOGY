```python
import pandas as pd

df = pd.read_csv('/content/placementdata.csv')

print(df.head())
```

```python
print(df.isnull().sum())
```

```python
df = df.dropna()

print("Missing values removed successfully")
```

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

df['ExtracurricularActivities'] = encoder.fit_transform(df['ExtracurricularActivities'])

df['PlacementTraining'] = encoder.fit_transform(df['PlacementTraining'])

df['PlacementStatus'] = encoder.fit_transform(df['PlacementStatus'])

print(df.head())
```

```python
import matplotlib.pyplot as plt

df.hist(figsize=(12,10))

plt.show()
```

```python
import seaborn as sns

plt.figure(figsize=(10,8))

sns.heatmap(df.corr(), annot=True)

plt.show()
```
